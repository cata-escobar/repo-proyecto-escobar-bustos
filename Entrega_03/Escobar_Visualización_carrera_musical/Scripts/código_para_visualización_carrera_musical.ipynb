```python
# Importamos Pandas y Altair en el notebook

import pandas as pd
import altair as alt

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 200)
```

```python
# Cargamos la base de datos

df = pd.read_csv("base_carrera_musical_3.csv", delimiter=";", encoding="UTF-8")
df.head()
```

# **Primer gráfico: Evolución por año de la cantidad de discos lanzados**

La primera visualización que realicé fue un gráfico de líneas desde un .groupby de la columna "año_lanzamiento", para que me mostrara una línea de tiempo de la cantidad de discos que habían sido lanzados en total por todos los proyectos estudiados.

En esta primera visualización, no hice diferenciación por tipo de disco o por banda/artista. Solo es un panorama general de la producción discográfica total en la ventana de tiempo analizada.

```python
por_anio_ordenado = (
    df.groupby("año_lanzamiento")
      .size()
      .reset_index(name="cantidad")
      .sort_values("año_lanzamiento")
)

por_anio_ordenado.head()

alt.Chart(por_anio_ordenado).mark_line(point=True).encode(
    x=alt.X("año_lanzamiento:O", title="Año"),
    y=alt.Y("cantidad:Q", title="Cantidad de discos"),
    tooltip=["año_lanzamiento", "cantidad"]
).properties(
    title="Cantidad de discos por año"
)
```

# **Segundo gráfico: filtración por banda**

Una vez que logré crear la visualización inicial, busqué poder diferenciar entre la producción discográfica de cada proyecto a lo largo de los años. Para esto, modifiqué el código original con un filtro al que llamé "bandas" y definí de acuerdo con los grupos que quería que me mostrara el gráfico.

Para elaborar este filtro tuve que pedirle ayuda a Gemini AI, ya que intenté definirlo de manera manual pero el código me arrojaba error. Me aseguré de que el código que me entregó Gemini AI fuera reproducible en el notebook y se ajustara a lo que yo necesitaba para este gráfico, modificándolo para que así fuera.

Además, definí un .groupby al que llamé "evolución", que agrupó las columnas "artista_proyecto" y "año_lanzamiento" de mi base de datos para asegurar la correcta interpretación de los datos por el código.

El resultado fue un gráfico que me permitió visualizar una comparación entre Blur, Oasis, Gorillaz y Nole Gallagher's High Flying Birds, los cuatro proyectos más relevantes del estudio.

```python
# Filtrar por banda
bandas = df[df["artista_proyecto"].isin(["Blur", "Oasis", "Gorillaz", "Noel Gallagher's High Flying Birds"])]

# Contar discos
evolucion = (
    bandas.groupby(["artista_proyecto", "año_lanzamiento"])
    .size()
    .reset_index(name="cantidad")
)

# Crear gráfico de líneas
grafico_2 = alt.Chart(evolucion).mark_line(point=True).encode(
    x=alt.X("año_lanzamiento:O", title="Año"),
    y=alt.Y("cantidad:Q", title="Cantidad de discos"),
    color=alt.Color("artista_proyecto:N", title="Banda"),
    tooltip=["artista_proyecto", "año_lanzamiento", "cantidad"]
).properties(
    title="Cantidad de discos por año por banda"
).interactive()

grafico_2
```

# **Tercer gráfico: Visualización de la totalidad de la discografía, diferenciada por bandas**

Cuando logré definir el código que me permitiera visualizar sin problemas el gráfico filtrado por bandas, quise ir más allá e incluir todos los grupos que figuraban en la columa "artista_proyecto" de la base de datos.

El resultado fue un gráfico de líneas con muchos colores, ninguno definido por mí, y sin una armonía visual.

```python
# Filtrar por banda
bandas = df[df["artista_proyecto"].isin(["Blur", "Oasis", "Gorillaz", "Noel Gallagher's High Flying Birds", "Noel Gallagher", "Liam Gallagher", "Damon Albarn", "Beady Eye", "The Good the Bad & the Queen"])]

# Contar discos
evolucion = (
    bandas.groupby(["artista_proyecto", "año_lanzamiento"])
    .size()
    .reset_index(name="cantidad")
)

# Crear gráfico de líneas
grafico_3 = alt.Chart(evolucion).mark_line(point=True).encode(
    x=alt.X("año_lanzamiento:O", title="Año"),
    y=alt.Y("cantidad:Q", title="Cantidad de discos"),
    color=alt.Color("artista_proyecto:N", title="Banda"),
    tooltip=["artista_proyecto", "año_lanzamiento", "cantidad"]
).properties(
    title="Cantidad de discos por año por banda"
).interactive()

grafico_3
```

# **Cuarto gráfico: Experimentación con gráficos de barras para analizar tipología discográfica**

Como los gráficos de líneas solo me servían para hacer un análisis de evolución histórica, quise experimentar con otro tipo de gráfico que me permitiera analizar el dominio de cada banda según tipo discográfico. En otras palabras, quería responder a la pregunta: ¿quién ganó en cada tipo de discografía?

Para lograr esto, construí un gráfico de barras simple dedsde la librería de Altair, a partir de un .groupby al que llamé "conteo" y que agrupaba lasa columnas "artista_proyecto" y "tipo_discografia" de la base de datos.

Luego, utilizando el código disponible en el notebook de clases, le pedí ayuda a Gemini AI para modificar el gráfico y poder editar la altura y ancho de la visualización.

El resultado fue un gráfico de barras inicial que comparaba el desempeño en lanzamientos discográficos, por tipo de discografía, de las bandas Oasis, Blur, Gorillaz y Noel Gallagher's High Flying Birds.

```python
# Filtrar por banda
datos = df[df["artista_proyecto"].isin(["Blur", "Oasis", "Gorillaz", "Noel Gallagher's High Flying Birds"])]

# Contar la cantidad de discos por banda
conteo = (
    datos.groupby(["artista_proyecto", "tipo_discografia"])
    .size()
    .reset_index(name="cantidad")
)

# Crear el gráfico
barras_1 = alt.Chart(conteo).mark_bar().encode(
    x=alt.X("tipo_discografia:N", title="Tipo de disco lanzado"),
    y=alt.Y("cantidad:Q", title="Número de discos"),
    color=alt.Color("artista_proyecto:N", title="Banda"),
    xOffset="artista_proyecto:N",
    tooltip=["artista_proyecto", "tipo_discografia", "cantidad"]
).properties(
    title="Blur vs. Oasis: carrera musical",
    width=700,
    height=350
)

barras_1
```

Cuando el notebook ejecutó el gráfico y pude analizarlo, me di cuenta que los dos proyectos de Damon Albarn que aparecían en él (Blur y Gorillaz) superaban en cantidad de discos a los de los Gallagher (Oasis y High Flying Birds), en las cuatro tipologías más relevantes de la industria: Album, Live album, EP y Live EP.

Esto me hizo llegar a una conclusión preliminar, respondiendo a la pregunta inicial que me hice antes de elaborar el gráfico: Damon Albarn le ganó a los Gallagher en cantidad de discos lanzados, en los 35 años que considera el estudio.

# Parte 1

Como en un principio partí trabajando con gráficos de líneas, decidí que mi historia iba a seguir por ese lado. Quise responder en parte a la hipótesis del trabajo de investigación, y lograr demostrar con una visualización de evolución temporal que Damon Albarn es quien habría logrado consolidarse tras 35 años de carrera musical.

Para eso, me guié del tercer gráfico que construí y decidí, primero modificar el tipo de gráfico para que fuera visualmente más accesible. Para eso, importé desde la biblioteca de Vega-Altair el código para el gráfico de líneas con marcadores de puntos, para que cada punto fuera mucho más visible en la visualización.

```python
chart = alt.Chart(evolucion).mark_line(
    point=alt.OverlayMarkDef(filled=False, fill="white")
).encode(
    x=alt.X("año_lanzamiento:O", title="Año"),
    y=alt.Y("cantidad:Q", title="Cantidad de discos"),
    color=alt.Color(
        "artista_proyecto:N",
        title="Banda",
    ),
    tooltip=["artista_proyecto", "año_lanzamiento", "cantidad"]
).properties(
    title="Cantidad de discos por año por banda"
).interactive()

chart
```

Una vez que tuve la visualización hecha, le pedí ayuda a Gemini AI para asignarle colores personalizados a cada variable (Banda). Así obtuve un código personalizable que me permitió establecer una clasificación que agrupara los proyectos de Damon Albarn bajo un grupo de colores inclinado hacia el amarillo/anaranjado y los de los Gallagher en otro grupo de colores, inclinado hacia los azules/grises.

```python
# Define los nombres exactos que aparecen en tu columna "artista_proyecto"
nombres_bandas = ["Beady Eye", "Blur", "Damon Albarn", "Gorillaz", "Liam Gallagher", "Noel Gallagher", "Noel Gallagher's High Flying Birds", "Oasis", "The Good the Bad & the Queen"]

# Define los colores que quieres para cada banda (en el mismo orden)
colores_personalizados = ["#6D8FAD", "#FFCC00", "#997A00", "#FF7F00", "#2A86D6", "#185182", "#3A6A93", "#006ECE", "#FDE92B"]

chart_2 = alt.Chart(evolucion).mark_line(
    point=alt.OverlayMarkDef(filled=False, fill="white")
).encode(
    x=alt.X("año_lanzamiento:O", title="Año"),
    y=alt.Y("cantidad:Q", title="Cantidad de discos"),
    color=alt.Color(
        "artista_proyecto:N",
        title="Banda",
        # Agrega la escala de colores aquí
        scale=alt.Scale(domain=nombres_bandas, range=colores_personalizados)
    ),
    tooltip=["artista_proyecto", "año_lanzamiento", "cantidad"]
).properties(
    title="Cantidad de discos por año por banda"
).interactive()

chart_2
```

# Parte 2

Cuando tuve la visualización de líneas lista, con los colores personalizados, me di cuenta que las líneas que se cruzaban en el gráfico generaban mucho ruido visual a la hora de leerlo, y hacían difícil poder diferenciar las evoluciones de cada banda por separado.

Por eso, decidí buscar en la biblioteca de Vega-Altair un gráfico de puntos, ya que, desde mi punto de vista, este iba a generar una visualización mucho más limpia y facilitaría la lectura.

Finalmente, me decidí por un gráfico simple de scatter plot con etiquetas, pero eliminé la parte del código que añadía texto en cada punto porque no lo necesitaba para efectos de mi visualización. Conservé el filtro por banda y el .groupby de "evolución", además de la definición de los colores personalizados para que se mantuviera la identidad visual de la visualización.


```python
# Definimos un filtro para las bandas que utiizaremos en la visualización
bandas = df[df["artista_proyecto"].isin(["Blur", "Oasis", "Gorillaz", "Noel Gallagher's High Flying Birds", "Noel Gallagher", "Liam Gallagher", "Damon Albarn", "Beady Eye", "The Good the Bad & the Queen"])]

# Contamos los discos
evolucion = (
    bandas.groupby(["artista_proyecto", "año_lanzamiento"])
    .size()
    .reset_index(name="cantidad")
)

# Importamos el código de la visualización de Vega-Altair para el gráfico de scatter plot

source = pd.DataFrame({
    'x': [1, 3, 5, 7, 9],
    'y': [1, 3, 5, 7, 9],
    'label': ['A', 'B', 'C', 'D', 'E']
})

# Definimos los nombres exactos que aparecen en la columna "artista_proyecto" de la base de datos

nombres_bandas = ["Beady Eye", "Blur", "Damon Albarn", "Gorillaz", "Liam Gallagher", "Noel Gallagher", "Noel Gallagher's High Flying Birds", "Oasis", "The Good the Bad & the Queen"]

# Definimos los colores que aparecerán en el gráfico para cada banda

colores_personalizados = ["#6D8FAD", "#FFCC00", "#997A00", "#FF7F00", "#2A86D6", "#185182", "#3A6A93", "#006ECE", "#FDE92B"]

# Creamos el código para el gráfico de scatter plot (que extraímos de Vega-Altair), asignándolo a una variable

point_chart_discos = alt.Chart(evolucion).mark_point().encode(
    x=alt.X("año_lanzamiento:O", title="Año"),
    y=alt.Y("cantidad:Q", title="Cantidad de discos"),
    color=alt.Color(
        "artista_proyecto:N",
        title="Banda",
        scale=alt.Scale(domain=nombres_bandas, range=colores_personalizados)
    ),
    tooltip=["artista_proyecto", "año_lanzamiento", "cantidad"]
).properties(
    title="Damon Albarn vs. los Gallagher: Vigencia discográfica a lo largo de los años"
)


# Se ejecuta el gráfico

point_chart_discos
```

Esta versión de la visualización hace que la lectura e interpretación del gráfico sea mucho más fluida, pudiendo identificar claramente los colores de cada banda dentro del gráfico, y evitando confusiones al eliminarse las líneas que se cruzaban para todos lados en el anterior.

Este será el gráfico que utilizaré para presentar como Visualización 1 en la Entrega 3 del trabajo.

Para terminar con la visualización, guardé el gráfico en formato HTML dentro del notebook.

```python
# Guardar el gráfico en HTML

point_chart_discos.save("vigencia_discográfica.html")
```
