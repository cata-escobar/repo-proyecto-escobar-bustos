```python
# 1. Define la lista exacta de las bandas que quieres incluir en el gráfico
bandas_a_mostrar = ["Blur", "Oasis", "Gorillaz", "Noel Gallagher's High Flying Birds"] # Añade o quita bandas aquí

# 2. Filtrar el DataFrame original por tipo de discografía Y por la lista de bandas
discos = df[
    df["tipo_discografia"].isin(["Album", "Compilation", "EP", "Live album", "Live EP"]) &
    df["artista_proyecto"].isin(bandas_a_mostrar)
]

# 3. Contar la cantidad de discos por banda
conteo = (
    discos.groupby(["artista_proyecto", "tipo_discografia"])
    .size()
    .reset_index(name="cantidad")
)

# 4. Crear el gráfico con los datos ya filtrados
grafico_6 = alt.Chart(conteo).mark_bar().encode(
    x=alt.X("tipo_discografia:N", title="Tipo de disco lanzado"),
    y=alt.Y("cantidad:Q", title="Número de discos"),
    color=alt.Color("artista_proyecto:N", title="Banda"),
    xOffset="artista_proyecto:N",
    tooltip=["artista_proyecto", "tipo_discografia", "cantidad"]
).properties(
    title="Blur vs. Oasis: carrera musical",
    width=1000,
    height=350
).interactive()

grafico_6
```
