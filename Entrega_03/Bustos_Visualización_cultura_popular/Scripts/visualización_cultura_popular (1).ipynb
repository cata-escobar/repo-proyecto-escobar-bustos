# **Primer gráfico de análisis**

```python
# 1. Importamos la librería y la base de datos

import pandas as pd
import altair as alt

df = pd.read_csv(
    "base_cultura_popular_limpia.csv",
    delimiter=";",
    encoding="UTF-8"
)
```

```python
# 2. Filtramos las bandas que queremos visualizar
datos = df[df["artista_proyecto"].isin(["Blur", "Oasis", "Gorillaz","Noel Gallagher's High Flying Birds","Noel Gallagher", "Liam Gallagher", "Damon Albarn", "The Good, the Bad & the Queen"])]

# 3. Contamos las apariciones por banda y el tipo de contenido
conteo = (
    datos.groupby(["artista_proyecto", "tipo"])
    .size()
    .reset_index(name="apariciones")
)

# 4. Creamos un gráfico de barras simple
grafico = alt.Chart(conteo).mark_bar().encode(
    x=alt.X("tipo:N", title="Tipo de contenido"),
    y=alt.Y("apariciones:Q", title="Número de apariciones"),
    color=alt.Color("artista_proyecto:N", title="Banda"),
    xOffset="artista_proyecto:N",
    tooltip=["artista_proyecto", "tipo", "apariciones"]
).properties(
    title="Presencia en la cultura popular",
    width=550,
    height=350
)

grafico
```

Tras evaluar las apariciones de cada banda según el tipo de contenido, fue posible identificar que Oasis sacaba ventaja en todas las categorías.

Sin embargo, tomando en cuenta que el objetivo de estos análisis, abrió la duda sobre cómo es la distribución total de apariciones agrupando los proyectos de Damon Albarn y los de hermanos Gallagher en "equipos".

# **Segundo gráfico de análisis**

```python
# 1. Agrupamos los proyectos por trayectoria, divididos entre Damon Albarn y Los Gallagher
grupos = {
    "Blur": "Damon Albarn",
    "Damon Albarn": "Damon Albarn",
    "Gorillaz": "Damon Albarn",
    "The Good, the Bad & the Queen": "Damon Albarn",
    "Oasis": "Los Gallagher",
    "Beady Eye": "Los Gallagher",
    "Liam Gallagher": "Los Gallagher",
    "Noel Gallagher": "Los Gallagher",
    "Noel Gallagher's High Flying Birds": "Los Gallagher"
}

df["grupo"] = df["artista_proyecto"].map(grupos)

# 2. Conservamos los proyectos clasificados
datos = df.dropna(
    subset=["grupo", "artista_proyecto"]
).copy()

# 3. Contamos a modo resumen las apariciones de cada proyecto
resumen = (
    datos.groupby(["grupo", "artista_proyecto"])
    .size()
    .reset_index(name="apariciones")
)

# 4. Creamos el gráfico de barras apiladas
grafico2 = alt.Chart(resumen).mark_bar().encode(
    x=alt.X(
        "grupo:N",
        title="Trayectoria musical",
        sort=["Damon Albarn", "Los Gallagher"],
        axis=alt.Axis(labelAngle=0)
    ),
    y=alt.Y(
        "apariciones:Q",
        title="Total de apariciones",
        stack="zero"
    ),
    color=alt.Color(
        "artista_proyecto:N",
        title="Proyecto musical"
    ),
    order=alt.Order(
        "apariciones:Q",
        sort="descending"
    ),
    tooltip=[
        alt.Tooltip("grupo:N", title="Trayectoria"),
        alt.Tooltip("artista_proyecto:N", title="Proyecto"),
        alt.Tooltip("apariciones:Q", title="Apariciones")
    ]
).properties(
    title="Damon Albarn vs Los Gallagher: distribución de apariciones según proyecto",
    width=500,
    height=380
)

grafico2
```

Se logró observar que todos los proyectos de Damon Albarn acumulaban la mayoría de apaciones entre ambos equipos, donde la distribución era prácticamente repartida entre Blur y Gorillaz.

En cambio, los Gallagher, la mayor cantidad de apariciones que tenían se concentraban en Oasis.

Esto me llamó mucho la atención ya que, a pesar que entre Liam y Noel acumulan 4 proyectos alternos en total, Damon logró sacarles ventaja solo con Gorillaz, aunque contaba con un total de 3 proyectos alternos.

# **Tercer gráfico de análisis**

```python

# 1. Contamos las apariciones por grupo y tipo de contenido

comparacion = (
    df.groupby(["grupo", "tipo"])
    .size()
    .reset_index(name="apariciones")
)

print(comparacion.to_string(index=False))
```

```python
# 2. Creamos gráfico de barras simple para comparar las apariciones por equipos
grafico3 = alt.Chart(comparacion).mark_bar().encode(
    x=alt.X(
        "tipo:N",
        title="Tipo de contenido",
        sort=["Serie", "Pelicula", "Videojuego"]
    ),
    y=alt.Y(
        "apariciones:Q",
        title="Número de apariciones"
    ),
    xOffset=alt.XOffset("grupo:N"),
    color=alt.Color(
        "grupo:N",
        title="Trayectoria",
        scale=alt.Scale(
            domain=["Damon Albarn", "Los Gallagher"],
            range=["#8064A2", "#4776A8"]
        )
    ),
    tooltip=["grupo", "tipo", "apariciones"]
).properties(
    title="Dos legados del Britpop: música en series, películas y videojuegos",
    width=500,
    height=350
)

grafico3
```

Con esta observación, Damon Albarn logra sacar ventaja en series y videojuegos. Aunque los Gallagher se alzan sobre Albarn en las películas, la diferencia es mínima, acumulando solo 3 aparaciones más.

Esto significa que, en términos generales, Damon Albarn acumula más apariciones en prácticamente todos los productos culturales. O, dicho de otro modo, los Gallagher a penas logran superar a Albarn únicamente a través de las películas.

# **Gráfico tentativo e interactivo para conocer el detalle de distribución por equipo y producto cultural**

```python

#1. Agrupar los proyectos por trayectoria
grupos = {
    "Blur": "Damon Albarn",
    "Damon Albarn": "Damon Albarn",
    "Gorillaz": "Damon Albarn",
    "The Good, the Bad & the Queen": "Damon Albarn",
    "Oasis": "Los Gallagher",
    "Beady Eye": "Los Gallagher",
    "Liam Gallagher": "Los Gallagher",
    "Noel Gallagher": "Los Gallagher",
    "Noel Gallagher's High Flying Birds": "Los Gallagher"
}

df["grupo"] = df["artista_proyecto"].map(grupos)

# 2. Limpiar y estandarizar los tipos de obra
datos = df.dropna(
    subset=["grupo", "tipo", "artista_proyecto"]
).copy()

datos["tipo"] = datos["tipo"].astype(str).str.strip().str.lower()

tipos = {
    "serie": "Series",
    "series": "Series",
    "pelicula": "Películas",
    "película": "Películas",
    "peliculas": "Películas",
    "películas": "Películas",
    "videojuego": "Videojuegos",
    "videojuegos": "Videojuegos"
}

datos["tipo"] = datos["tipo"].replace(tipos)

datos = datos[
    datos["tipo"].isin(["Series", "Películas", "Videojuegos"])
].copy()

# 3. Contamos las apariciones por trayectoria, tipo y proyecto
resumen = (
    datos.groupby(["grupo", "tipo", "artista_proyecto"])
    .size()
    .reset_index(name="apariciones")
)

# 4. Creamos el selector del tipo de obra
selector_tipo = alt.selection_point(
    fields=["tipo"],
    bind=alt.binding_select(
        options=["Series", "Películas", "Videojuegos"],
        name="Tipo de obra: "
    ),
    value="Videojuegos"
)

# 5. Selector de trayectoria
selector_grupo = alt.selection_point(
    fields=["grupo"],
    bind=alt.binding_select(
        options=["Damon Albarn", "Los Gallagher"],
        name="Trayectoria: "
    ),
    value="Damon Albarn"
)

# 6. Creamos el gráfico interactivo
grafico4 = alt.Chart(resumen).mark_bar().encode(
    y=alt.Y(
        "artista_proyecto:N",
        title="Proyecto musical",
        sort="-x"
    ),
    x=alt.X(
        "apariciones:Q",
        title="Cantidad de apariciones"
    ),
    color=alt.Color(
        "artista_proyecto:N",
        title="Proyecto"
    ),
    tooltip=[
        alt.Tooltip("grupo:N", title="Trayectoria"),
        alt.Tooltip("tipo:N", title="Tipo de obra"),
        alt.Tooltip("artista_proyecto:N", title="Proyecto"),
        alt.Tooltip("apariciones:Q", title="Apariciones")
    ]
).add_params(
    selector_tipo,
    selector_grupo
).transform_filter(
    selector_tipo
).transform_filter(
    selector_grupo
).properties(
    title="¿Qué proyectos mantienen la presencia cultural de cada trayectoria?",
    width=1000,
    height=300
)

grafico4
```

Este gráfico fue tentativo para intentar poner todo en uno, por lo que tuve que emplear inteligencia artificial, ya que exigía que fuese interactivo y tuve que modificar muchas veces el tamaño.

Sin embargo, dado que las leyendas se cortan y siento que no parece tan ilustrativo para mostrar las diferencias en términos de apariciones, decidí no usarlo para esta entrega.
