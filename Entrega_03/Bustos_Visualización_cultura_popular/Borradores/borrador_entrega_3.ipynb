```python
import pandas as pd
import altair as alt

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
# Filtrar únicamente Blur y Oasis
datos = df[df["artista_proyecto"].isin(["Blur", "Oasis", "Gorillaz","Noel Gallagher's High Flying Birds","Noel Gallagher", "Liam Gallagher", "Damon Albarn"])]

# Contar las apariciones por banda y tipo de contenido
conteo = (
    datos.groupby(["artista_proyecto", "tipo"])
    .size()
    .reset_index(name="apariciones")
)

# Crear el gráfico
grafico = alt.Chart(conteo).mark_bar().encode(
    x=alt.X("tipo:N", title="Tipo de contenido"),
    y=alt.Y("apariciones:Q", title="Número de apariciones"),
    color=alt.Color("artista_proyecto:N", title="Banda"),
    xOffset="artista_proyecto:N",
    tooltip=["artista_proyecto", "tipo", "apariciones"]
).properties(
    title="Blur vs. Oasis: presencia en la cultura popular",
    width=500,
    height=350
)

grafico
```

En videojuegos, Damon Albarn obtiene mayor presencia con Blur y Gorrillaz. Mientras que en películas y series, Oasis gana más terreno.

```python
# Filtrar Blur y Oasis
datos = df[
    df["artista_proyecto"].isin(["Blur", "Oasis","Gorillaz"])
].dropna(subset=["aparicion"])

# Contar las apariciones por año y banda
evolucion = (
    datos.groupby(["aparicion", "artista_proyecto"])
    .size()
    .reset_index(name="apariciones")
)

# Crear gráfico de líneas
grafico = alt.Chart(evolucion).mark_line(
    point=True
).encode(
    x=alt.X(
        "aparicion:O",
        title="Año de aparición",
        axis=alt.Axis(labelAngle=-45)
    ),
    y=alt.Y(
        "apariciones:Q",
        title="Número de apariciones",
        scale=alt.Scale(zero=True)
    ),
    color=alt.Color(
        "artista_proyecto:N",
        title="Banda"
    ),
    tooltip=[
        alt.Tooltip("aparicion:O", title="Año"),
        alt.Tooltip("artista_proyecto:N", title="Banda"),
        alt.Tooltip("apariciones:Q", title="Apariciones")
    ]
).properties(
    title="Blur vs. Oasis: evolución de su presencia en la cultura popular",
    width=650,
    height=400
).interactive()

grafico
```
